```python
from pathlib import Path
import json
import time

import matplotlib.pyplot as plt
from IPython.display import clear_output

RUN_ID = 0
PROGRESS_PATH = Path("trained_models_patterns_bilobed_ring_contacts") / f"train-pattern-{RUN_ID}" / "training_progress.json"
```

```python
def plot_training_progress():
    with open(PROGRESS_PATH, "r") as f:
        rows = json.load(f)

    epochs = [row["epoch"] for row in rows]
    train_loss = [row["train_loss"] for row in rows]
    val_loss = [row.get("val_loss") for row in rows]
    has_val = any(v is not None for v in val_loss)
    j_values = [row.get("j") for row in rows]
    fractions = [row.get("type_fractions") for row in rows]
    counts = rows[-1].get("type_counts") if rows else None

    fig, axes = plt.subplots(1, 3, figsize=(18, 4))
    axes[0].plot(epochs, train_loss, label="train")
    if has_val:
        axes[0].plot(epochs, val_loss, label="val")
        axes[0].legend()
    axes[0].set_xlabel("epoch")
    axes[0].set_ylabel("loss")
    axes[0].grid(alpha=0.3)

    if j_values and j_values[0] is not None:
        labels = ["J11", "J12", "J13", "J22", "J23", "J33"]
        for idx, label in enumerate(labels):
            axes[1].plot(epochs, [j[idx] for j in j_values], label=label)
        axes[1].set_xlabel("epoch")
        axes[1].set_ylabel("visible J")
        axes[1].legend(fontsize=8, ncol=2)
        axes[1].grid(alpha=0.3)

    if fractions and fractions[0] is not None:
        labels = ["frac 1", "frac 2", "frac 3"]
        for idx, label in enumerate(labels):
            axes[2].plot(epochs, [frac[idx] for frac in fractions], label=label)
        axes[2].set_xlabel("epoch")
        axes[2].set_ylabel("learned fraction")
        axes[2].legend(fontsize=8)
        axes[2].grid(alpha=0.3)
        if counts is not None:
            axes[2].set_title(f"latest counts: {counts}")

    fig.tight_layout()
    plt.show()
```

```python
while True:
    clear_output(wait=True)
    plot_training_progress()
    time.sleep(10)
```
